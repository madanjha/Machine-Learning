Objective:
The primary objective is to predict the output value y based on two input features
x1 and x2 using a basic neural network model with the equation y=3x1+4x2. The
neural network will be trained using a feed-forward and back-propagation
mechanism to minimize the error between the predicted and actual values.
Methodology
1. Activation Functions:
● Define the sigmoid function, a popular activation function used in neural
networks.
● Define the tanh function, which is another common activation function that
returns values between -1 and 1.
● Define the relu function, which returns positive values as they are and
zeroes out negative values
2. Sample data
   use the following sample
   x1=3,x2=2, yactual=17
3. Feed forward
   Initialize the random weights w1 and w2 between 1 and 10.
   Predict the output ypred using the equation ypred= x1 x w1 + x2 x w2
4. Error Calculation:
  ● Compute the squared error as error = (yactual-ypred)*(yactual-ypred)
5. Back Propagation:
  ● Compute the gradients of the error with respect to the weights.
  ● Update the weights using the computed gradients and a learning rate.
6. Training:
  ● The above steps (Feed Forward to Back Propagation) are performed
  iteratively (20 times in this example) to refine the weights and minimize
  the error.
7. Visualization:
  ● Plot the progression of the error across epochs.
  ● Plot the progression of ypred across epochs.
Expected Outcome:
  After sufficient training, the predicted ypred should be close to the actual value
  yactual=17. The error graph should show a declining trend, indicating that the
  model is learning and improving its predictions over time.

```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def tanh(x):
    return math.tanh(x)

def relu(x):
    return max(0, x)

print("Activation functions (sigmoid, tanh, relu) defined.")
```

```python
x1 = 3
x2 = 2
y_actual = 17

print(f"Sample data prepared: x1={x1}, x2={x2}, y_actual={y_actual}")
```

```python
import random

# Initialize weights randomly between 1 and 10
w1 = random.uniform(1, 10)
w2 = random.uniform(1, 10)

# Calculate ypred using the feed-forward equation
ypred = x1 * w1 + x2 * w2

print(f"Initial weights: w1={w1:.4f}, w2={w2:.4f}")
print(f"Initial ypred: {ypred:.4f}")
```

Error Calculation

```python
error = (y_actual - ypred) * (y_actual - ypred)

print(f"Squared Error: {error:.4f}")
```

Back Propagation

```python
learning_rate = 0.01

# Calculate gradients
gradient_w1 = -2 * (y_actual - ypred) * x1
gradient_w2 = -2 * (y_actual - ypred) * x2

# Update weights
w1 = w1 - learning_rate * gradient_w1
w2 = w2 - learning_rate * gradient_w2

print(f"Updated w1: {w1:.4f}")
print(f"Updated w2: {w2:.4f}")
print(f"Gradient w1: {gradient_w1:.4f}")
print(f"Gradient w2: {gradient_w2:.4f}")
```

Training Loop

```python
import random

# Re-initialize weights randomly at the start of the training loop
w1 = random.uniform(1, 10)
w2 = random.uniform(1, 10)

learning_rate = 0.01
epochs = 20

errors_per_epoch = []
ypreds_per_epoch = []

print(f"Initial weights for training loop: w1={w1:.4f}, w2={w2:.4f}\n")

for epoch in range(epochs):
    # 1. Feed-forward
    ypred = x1 * w1 + x2 * w2
    
    # 2. Error Calculation
    error = (y_actual - ypred) * (y_actual - ypred)
    
    # Store for visualization
    errors_per_epoch.append(error)
    ypreds_per_epoch.append(ypred)
    
    # 3. Back Propagation
    # Calculate gradients
    gradient_w1 = -2 * (y_actual - ypred) * x1
    gradient_w2 = -2 * (y_actual - ypred) * x2
    
    # Update weights
    w1 = w1 - learning_rate * gradient_w1
    w2 = w2 - learning_rate * gradient_w2
    
    # Print progress for each epoch (optional, but good for tracking)
    print(f"Epoch {epoch + 1}/{epochs}: ypred={ypred:.4f}, Error={error:.4f}, w1={w1:.4f}, w2={w2:.4f}")

print("\n--- Training Complete ---")
print(f"Final weights: w1={w1:.4f}, w2={w2:.4f}")
print(f"Final ypred: {ypreds_per_epoch[-1]:.4f}")
print(f"Final Error: {errors_per_epoch[-1]:.4f}")
```

```python
import matplotlib.pyplot as plt

epochs_list = range(1, epochs + 1)

plt.figure(figsize=(12, 5))

# Plotting Error Progression
plt.subplot(1, 2, 1)
plt.plot(epochs_list, errors_per_epoch, marker='o', linestyle='-', color='blue')
plt.title('Error Progression Across Epochs')
plt.xlabel('Epoch')
plt.ylabel('Squared Error')
plt.grid(True)
plt.legend(['Squared Error'])

# Plotting ypred Progression
plt.subplot(1, 2, 2)
plt.plot(epochs_list, ypreds_per_epoch, marker='x', linestyle='--', color='red', label='Predicted y')
plt.axhline(y=y_actual, color='green', linestyle='-', label='Actual y')
plt.title('Predicted y Progression Across Epochs')
plt.xlabel('Epoch')
plt.ylabel('Predicted y Value')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()

print("Visualizations generated for error and ypred progression.")
```
