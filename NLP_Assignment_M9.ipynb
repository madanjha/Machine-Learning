## Preprocess Text Data

### Subtask:
Clean the text data by removing stop words, punctuation, converting text to lowercase, and applying any other necessary text cleaning steps.


**Reasoning**:
I will import the necessary libraries, download the NLTK stopwords, define the `preprocess_text` function to clean the text data, and then apply this function to the 'text' column of `filtered_df` to create a new 'cleaned_text' column. Finally, I will display the first few rows of the updated DataFrame to verify the cleaning process.



```python
import re
import nltk
from nltk.corpus import stopwords

# Download NLTK stopwords if not already downloaded
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')

# Get English stopwords
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    # Convert to lowercase
    text = text.lower()
    # Remove punctuation
    text = re.sub(r'[^a-z0-9\s]', '', text)
    # Remove numbers
    text = re.sub(r'\d+', '', text)
    # Remove stop words
    text = ' '.join(word for word in text.split() if word not in stop_words)
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Apply the preprocessing function to the 'text' column
filtered_df['cleaned_text'] = filtered_df['text'].apply(preprocess_text)

# Display the first few rows with original and cleaned text
print("DataFrame with original and cleaned text:")
print(filtered_df[['text', 'cleaned_text']].head())
```

**Reasoning**:
The previous step successfully cleaned the text data. The next logical step, as per the problem statement, is to vectorize this preprocessed text data. This involves tokenization and padding to prepare it for a deep learning model.



```python
## Text Vectorization

### Subtask:
Convert the preprocessed text data ('cleaned_text' column) into a numerical format using tokenization and padding, making it suitable for a deep learning model.

#### Instructions:
1.  Initialize a `Tokenizer` from `tensorflow.keras.preprocessing.text` with a specified vocabulary size (e.g., `num_words=10000`) and configure it to handle out-of-vocabulary (OOV) words by setting `oov_token="<unk>"`.
2.  Fit the `Tokenizer` on the 'cleaned_text' column of the `filtered_df` to build the vocabulary.
3.  Convert the cleaned text into sequences of integers using `texts_to_sequences()`.
4.  Determine a suitable maximum sequence length for padding. This could be the length of the longest sequence or a chosen fixed length.
5.  Pad the sequences to the `max_sequence_length` using `pad_sequences` from `tensorflow.keras.preprocessing.sequence`. Set `padding='post'` and `truncating='post'`.
6.  Print the shape of the padded sequences and display the first padded sequence to verify the vectorization.
```

**Reasoning**:
The previous attempt resulted in a `SyntaxError` because markdown content was placed in a code cell. The intent was to implement text vectorization. I will now provide the correct Python code to perform text vectorization, following the instructions previously outlined for tokenization and padding using TensorFlow Keras.



```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Initialize a Tokenizer
# A reasonable vocabulary size, can be adjusted based on data size and desired coverage
num_words = 10000
oov_token = "<unk>"
tokenizer = Tokenizer(num_words=num_words, oov_token=oov_token)

# 2. Fit the Tokenizer on the 'cleaned_text' column
tokenizer.fit_on_texts(filtered_df['cleaned_text'])

# 3. Convert the cleaned text into sequences of integers
text_sequences = tokenizer.texts_to_sequences(filtered_df['cleaned_text'])

# 4. Determine a suitable maximum sequence length for padding
# Using the length of the longest sequence in the current data
max_sequence_length = max([len(seq) for seq in text_sequences])

print(f"Maximum sequence length for padding: {max_sequence_length}")

# 5. Pad the sequences to the max_sequence_length
padded_sequences = pad_sequences(text_sequences,
                                 maxlen=max_sequence_length,
                                 padding='post', # Pad with zeros at the end
                                 truncating='post') # Truncate from the end if sequence is longer than maxlen

# 6. Print the shape of the padded sequences and display the first padded sequence
print("\nShape of padded sequences:", padded_sequences.shape)
print("First padded sequence:\n", padded_sequences[0])

# Store the tokenizer and padded sequences in variables that can be accessed later
# This makes them globally available in the notebook session.
tokenizer_nlp = tokenizer
X_padded = padded_sequences
```

**Reasoning**:
The text data has been vectorized. The next step is to prepare the target variable, 'sentiment', by encoding it into a numerical format suitable for a deep learning model, specifically one-hot encoding for a multi-class classification problem as indicated by the 'three categories' requirement.



```python
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.utils import to_categorical

# 1. Encode the sentiment labels to numerical values
label_encoder = LabelEncoder()
filtered_df['sentiment_encoded'] = label_encoder.fit_transform(filtered_df['sentiment'])

# Get the number of unique sentiment categories
num_classes = len(label_encoder.classes_)

# 2. One-hot encode the numerical sentiment labels
y_encoded = to_categorical(filtered_df['sentiment_encoded'], num_classes=num_classes)

print("Original sentiment labels:")
print(filtered_df['sentiment'].head())
print("\nEncoded sentiment labels (numerical):")
print(filtered_df['sentiment_encoded'].head())
print("\nOne-hot encoded sentiment labels (first 5 rows):")
print(y_encoded[:5])
print("\nShape of one-hot encoded labels:", y_encoded.shape)

# Store the label encoder for future use (e.g., inverse transform for predictions)
label_encoder_nlp = label_encoder
```

**Reasoning**:

```python
from sklearn.model_selection import train_test_split

# Split the data into training and validation sets
# Using a 80/20 split for training and validation
X_train, X_val, y_train, y_val = train_test_split(X_padded, y_encoded, test_size=0.2, random_state=42, stratify=filtered_df['sentiment_encoded'])

print("Shape of X_train:", X_train.shape)
print("Shape of X_val:", X_val.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_val:", y_val.shape)
```

**Reasoning**:
The data is now preprocessed, vectorized, and split into training and validation sets. The next step is to develop the deep learning model as outlined in the problem statement, which specifies an Embedding layer, a SpatialDropout1D layer, an LSTM layer, and a Dense layer for sentiment classification.



```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SpatialDropout1D, LSTM, Dense

# Model parameters (from previous steps)
vocab_size = num_words # num_words from Tokenizer initialization
embedding_dim = 100 # A common choice for embedding dimensions
max_length = max_sequence_length # max_sequence_length from padding
output_dim = num_classes # num_classes from one-hot encoding

# 1. Initialize the Sequential model
model = Sequential()

# 2. Add Embedding layer
# input_dim: size of the vocabulary (vocab_size)
# output_dim: dimension of the dense embedding (embedding_dim)
# input_length: length of input sequences (max_length)
model.add(Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_length))

# 3. Add SpatialDropout1D layer to prevent overfitting
# dropout rate of 0.2-0.5 is common
model.add(SpatialDropout1D(0.2))

# 4. Add LSTM layer for sequence data processing
# units: dimensionality of the output space (e.g., 100 or 128)
model.add(LSTM(128))

# 5. Add Dense output layer with softmax activation for multi-class classification
model.add(Dense(output_dim, activation='softmax'))

# Compile the model
# optimizer: 'adam' is a good general-purpose optimizer
# loss: 'categorical_crossentropy' for one-hot encoded labels
# metrics: 'accuracy' for evaluation
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Print model summary to verify the architecture
print(model.summary())
```

**Reasoning**:
The previous code produced a `UserWarning` indicating that the `input_length` argument for the Embedding layer is deprecated. Removing this argument will resolve the warning and allow the model to build its input shape dynamically when data is first passed during training.



```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SpatialDropout1D, LSTM, Dense

# Model parameters (from previous steps)
vocab_size = num_words # num_words from Tokenizer initialization
embedding_dim = 100 # A common choice for embedding dimensions
max_length = max_sequence_length # max_sequence_length from padding
output_dim = num_classes # num_classes from one-hot encoding

# 1. Initialize the Sequential model
model = Sequential()

# 2. Add Embedding layer
# input_dim: size of the vocabulary (vocab_size)
# output_dim: dimension of the dense embedding (embedding_dim)
# input_length: length of input sequences (max_length) - REMOVED as deprecated
model.add(Embedding(input_dim=vocab_size, output_dim=embedding_dim)) # Removed input_length

# 3. Add SpatialDropout1D layer to prevent overfitting
# dropout rate of 0.2-0.5 is common
model.add(SpatialDropout1D(0.2))

# 4. Add LSTM layer for sequence data processing
# units: dimensionality of the output space (e.g., 100 or 128)
model.add(LSTM(128))

# 5. Add Dense output layer with softmax activation for multi-class classification
model.add(Dense(output_dim, activation='softmax'))

# Compile the model
# optimizer: 'adam' is a good general-purpose optimizer
# loss: 'categorical_crossentropy' for one-hot encoded labels
# metrics: 'accuracy' for evaluation
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Print model summary to verify the architecture
print(model.summary())
```

**Reasoning**:
The model architecture has been defined and compiled. The next step, as per the problem statement, is to train the model using the prepared training and validation datasets. I will use the fit method, specifying the training data, validation data, epochs, and batch size.



```python
epochs = 10 # Number of training epochs
batch_size = 1 # Small batch size due to small dataset

print("Starting model training...")

history = model.fit(X_train,
                    y_train,
                    epochs=epochs,
                    batch_size=batch_size,
                    validation_data=(X_val, y_val),
                    verbose=1)

print("Model training finished.")

# Optionally, print the history keys to see what metrics were tracked
print("\nTraining history keys:", history.history.keys())
```

**Reasoning**:
The model has been trained. The next step is to evaluate its performance on the validation dataset to assess its accuracy and loss, as specified in the problem statement.



```python
print("Evaluating model on validation data...")

loss, accuracy = model.evaluate(X_val, y_val, verbose=0)

print(f"Validation Loss: {loss:.4f}")
print(f"Validation Accuracy: {accuracy:.4f}")
```
