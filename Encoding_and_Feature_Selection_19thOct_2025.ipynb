##**Encoding and it's type**
When we try to convert the categorical data into numerical.

There are three types of encoding:
1. **Label Encoding**: It converts the categorical data into numerical without any specific order.
2. **Ordinal Encoding**: It Converts the categories into numerical but in passed order (It is useful when we have ordinal data)
3. **One Hot Encoding**: It Creates a new binary columns(two values: 1 or 0) for each category, removing any implied order which makes it suitable for nominal Data

```python
import pandas as pd
data = {
    "Color":['Red','Blue','Red','Black'] ,
    "Size": ['Large', 'Small','Medium','Large'],
    "Marital Status" : ['Single','Married','Relationship','Single']

}
```

```python
data
```

```python
df = pd.DataFrame(data)
df
```

**Label Encoding** (Follows the alphabetical order)

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Color'] = le.fit_transform(df['Color'])
df
```

**Ordinal Encoding**

```python
from sklearn.preprocessing import OrdinalEncoder
'''
small = 0
medium = 1
large = 2
'''

oe = OrdinalEncoder(categories=[['Small','Medium','Large']]) # we are manually defining the order [0,1,2]
df['Size'] = oe.fit_transform(df[['Size']])
df
```

**One Hot Encoding**


```python
df = pd.get_dummies(df,columns=['Marital Status'])
df
```

##**Dimensionality Reduction Technique**
Let say we have a big table of data with many columns where, each column represents differnet informations about something. Sometimes, not all columns are useful.

**Dimensionality Reduction** is the process of reducing the number of columns while keeping the important information. It help us to do calculation faster and also makes model simpler so that we can get increased accuracy

**There are Two approches we can follow**
###**1. Feature Selection**
Feature Selection is like picking the most important columns from the table and ignoring the rest.
* **VIF (Variance Inflation Factor)**: Helps remove features that are very similar to each other (multicolinearity). so that our model does not get confused.
* **RFE (Recursive Feature Elimination)**: It tests features one by one to see which are the most important and removes the less usefull ones.

###**2. Feature Extraction**
It is a technique to create new features by combining or transforming the exisitng ones. Instead of just picking columns, it makes new columns that summarize the original data.
* **PCA (Principal Component Analysis)**: It Combines the column into fewer new ones that capture the most information
* **LDA (Linear Discriminant Analysis)**: It creates new features to better seperate different groups in the data.


**Conclusion**

Both Reduces dimensions and make data easuer to work with and model more effiecient

**Let's implement VIF**

**We first drop all the null values**

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

**Now if we have to select all numerical columns**

```python
numericCols = df.select_dtypes(include=np.number)
numericCols
```

**Finally We will implement VIF**

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif
```

```python
thresh = 5

while True:
  vifData = pd.DataFrame()
  vifData['Feature'] = numericCols.columns
  vifData['VIF'] = [vif(numericCols.values, i) for i in range(len(numericCols.columns))]
  max_vif = vifData['VIF'].max()
  if max_vif > thresh:
    max_feature = vifData.loc[vifData['VIF'].idxmax(),'Feature']
    print(f'Removing {max_feature} with vif = {max_vif:.2f}')
    numericCols = numericCols.drop(columns = [max_feature])
  else:
    break


print(f'Final Feature with acceptable VIF')
vifData
```

**Implementaion of RFE**

```python
import numpy as np
df = pd.read_csv('/content/bank-additional-full.csv',sep=';')
df
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

**We need to encode first to pass the data**

```python
le = LabelEncoder()
for i in df.columns:
  if df[i].dtypes == 'object':
    df[i] = le.fit_transform(df[i])
```

```python
df.head()
```

```python
df.info()
```

**Now let's start building model for RFE**

```python
x = df.drop("y",axis= 1)
y = df['y']
```

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
rfe = RFE(model) # passing the model inside rfe to apply RFE on it
rfe.fit(x,y)
```

**RFE will give us the index of columns that are important**

```python
select_index =  [i for i, data in enumerate(rfe.support_) if data == True]
```

**Store the column name in a list**

```python
column_name = x.columns
```

```python
selected_column_names = [column_name[i] for i in select_index]
selected_column_names
```

Now we will select the columns and check the scores

```python
x = df.loc[:,['marital',
 'default',
 'contact',
 'month',
 'day_of_week',
 'campaign',
 'previous',
 'poutcome',
 'cons.conf.idx',
 'euribor3m']]

y = df['y']
```

Data splitting into train and test set

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

**Now the model will use for prediction**

```python
LogReg = LogisticRegression()
```

```python
LogReg.fit(x_train,y_train)
```

**Test our Model**

```python
y_pred = LogReg.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import *
```

```python
print(f'Accuracy score of Model After applying RFE: {accuracy_score(y_pred,y_test)*100:.2f}%')
```

```python
print(classification_report(y_pred,y_test))
```

```python
def predictFunc(*args):
  data_dict = {col: [val] for col,val in zip(x.columns,args)}
  input_data = pd.DataFrame(data_dict)

  pred = LogReg.predict(input_data)[0]
  prob = LogReg.predict_proba(input_data)[0][1]

  return f'Prediction : {"Yes" if  pred ==1 else "No"} (probabilty of "Yes" = {prob:.2f})'
```
