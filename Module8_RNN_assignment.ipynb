## Load and Preprocess Data

### Subtask:
Load the IMDB dataset, keeping only the top 10,000 most frequently occurring words. Pad the sequences to ensure they all have the same length, preparing them for input into the RNN model.


**Reasoning**:
First, I'll import the necessary TensorFlow and Keras modules to load and preprocess the IMDB dataset, which is a prerequisite for the subsequent data loading and padding steps.



```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

print("TensorFlow and Keras modules imported successfully.")
```

**Reasoning**:
Now that the necessary libraries are imported, I will load the IMDB dataset, keeping only the top 10,000 most frequent words. I will then calculate an appropriate maximum sequence length (90th percentile of review lengths) and pad both the training and test sequences to this length, preparing them for model input.



```python
max_features = 10000
print(f"Loading IMDB dataset with top {max_features} words...")
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

print(f"Original training sequence shape: {x_train.shape}")
print(f"Original testing sequence shape: {x_test.shape}")

# Determine an appropriate maximum sequence length
train_lengths = [len(x) for x in x_train]
max_len = int(np.percentile(train_lengths, 90)) # Use 90th percentile for maxlen

print(f"Calculated maximum sequence length (90th percentile): {max_len}")

# Pad sequences to the chosen maximum length
x_train = pad_sequences(x_train, maxlen=max_len, padding='post', truncating='post')
x_test = pad_sequences(x_test, maxlen=max_len, padding='post', truncating='post')

print(f"Padded training sequence shape: {x_train.shape}")
print(f"Padded testing sequence shape: {x_test.shape}")
print("IMDB dataset loaded and sequences padded successfully.")
```

## Build RNN Model

### Subtask:
Construct a sequential Keras model consisting of an Embedding layer, a Simple RNN layer, and a Dense output layer for binary classification (positive/negative sentiment).


**Reasoning**:
I will import the required Keras layers and `Sequential` model to define the RNN architecture as specified, including an Embedding layer, a SimpleRNN layer, and a Dense output layer, then display the model summary to confirm its structure.



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

# Define the RNN model
embedding_dim = 128 # Output dimension for the embedding layer
rnn_units = 128 # Units for the SimpleRNN layer

model = Sequential([
    Embedding(input_dim=max_features, output_dim=embedding_dim, input_length=max_len),
    SimpleRNN(units=rnn_units),
    Dense(1, activation='sigmoid')
])

print("RNN model created successfully.")
model.summary()
```

**Reasoning**:
The previous attempt to build the model resulted in an 'unbuilt' model and a UserWarning about the deprecated `input_length` argument. To ensure the model is built correctly and parameters are displayed, I will remove the `input_length` argument and instead specify the `input_shape` directly in the Embedding layer, which is the recommended practice for newer Keras versions.



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

# Define the RNN model
embedding_dim = 128 # Output dimension for the embedding layer
rnn_units = 128 # Units for the SimpleRNN layer

model = Sequential([
    Embedding(input_dim=max_features, output_dim=embedding_dim, input_shape=(max_len,)), # Use input_shape instead of input_length
    SimpleRNN(units=rnn_units),
    Dense(1, activation='sigmoid')
])

print("RNN model created successfully.")
model.summary()
```

**Reasoning**:
To comply with the latest Keras recommendations and avoid the `UserWarning` regarding `input_shape` in `Sequential` models, I will explicitly define an `Input` layer as the first layer in the model. This clarifies the input shape for the model without implicitly passing it to the `Embedding` layer.



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras import Input

# Define the RNN model
embedding_dim = 128 # Output dimension for the embedding layer
rnn_units = 128 # Units for the SimpleRNN layer

model = Sequential([
    Input(shape=(max_len,)), # Explicitly define Input layer
    Embedding(input_dim=max_features, output_dim=embedding_dim), # No need for input_shape/input_length here
    SimpleRNN(units=rnn_units),
    Dense(1, activation='sigmoid')
])

print("RNN model created successfully.")
model.summary()
```

**Reasoning**:
Now that the RNN model architecture is defined, I will compile it. This involves specifying the appropriate optimizer (Adam), loss function (binary cross-entropy for binary classification), and metrics (accuracy) as required for training a sentiment analysis model.



```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
print("RNN model compiled successfully.")
```

## Train Model

### Subtask:
Train the compiled model on the preprocessed movie review data. Use a batch size of 128 and allocate 20% of the training data for validation. The training process should run for 10 epochs.


**Reasoning**:
I will train the compiled RNN model using the specified training data, batch size, number of epochs, and validation split, storing the training history in a variable.



```python
batch_size = 128
epochs = 10

print(f"Starting model training with batch_size={batch_size}, epochs={epochs}, and validation_split=0.2...")
history = model.fit(x_train, y_train, batch_size=batch_size, epochs=epochs, validation_split=0.2)

print("Model training completed. Training history stored.")
```

**Reasoning**:
Now that the model has been trained, I need to evaluate its performance on the unseen test dataset to measure its accuracy.



```python
print("Evaluating the model on the test set...")
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)

print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
```
