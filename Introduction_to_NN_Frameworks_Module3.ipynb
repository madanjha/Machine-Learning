## Load and Preprocess Data

### Subtask:
Load the MNIST dataset, normalize image pixel values to a [0, 1] range, and convert labels to one-hot encoded format. Ensure data is reshaped appropriately for the neural network input.


**Reasoning**:
To begin the data preprocessing, I will load the MNIST dataset, normalize the image pixel values, reshape the images for network input, and convert the labels to one-hot encoding, all within a single code block for efficiency.



```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# 1. Load the MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

print(f"Original training images shape: {x_train.shape}")
print(f"Original testing images shape: {x_test.shape}")
print(f"Original training labels shape: {y_train.shape}")
print(f"Original testing labels shape: {y_test.shape}")

# 2. Normalize pixel values
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# 3. Reshape the image data
# For a simple dense network, images are flattened from 28x28 to 784
num_pixels = x_train.shape[1] * x_train.shape[2]
x_train = x_train.reshape(x_train.shape[0], num_pixels)
x_test = x_test.reshape(x_test.shape[0], num_pixels)

print(f"Reshaped training images shape: {x_train.shape}")
print(f"Reshaped testing images shape: {x_test.shape}")

# 4. Convert labels to one-hot encoded format
num_classes = 10
y_train = to_categorical(y_train, num_classes)
y_test = to_categorical(y_test, num_classes)

print(f"One-hot encoded training labels shape: {y_train.shape}")
print(f"One-hot encoded testing labels shape: {y_test.shape}")

print("MNIST dataset loaded, normalized, reshaped, and labels one-hot encoded successfully.")
```

## Define Neural Network Model

### Subtask:
Construct a neural network model with an input layer suitable for 28x28 grayscale images, one or more hidden layers with ReLU activation, and an output layer with 10 units and softmax activation for classification.


**Reasoning**:
To construct the neural network model as specified, I will import the necessary Keras components, define a Sequential model, and add the input, hidden, and output Dense layers with the correct activation functions and unit counts.



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 1. Instantiate a Sequential model
model = Sequential()

# 2. Add a Dense layer as the input layer
# num_pixels was defined in the previous step as 784
model.add(Dense(128, activation='relu', input_shape=(num_pixels,)))

# 3. Add at least one more Dense hidden layer
model.add(Dense(64, activation='relu'))

# 4. Add a final Dense output layer
# num_classes was defined in the previous step as 10
model.add(Dense(num_classes, activation='softmax'))

# Print the model summary to confirm its structure
print("Neural Network Model Summary:")
model.summary()

print("Neural network model constructed successfully.")
```

**Reasoning**:
The previous output included a Keras UserWarning regarding the `input_shape` argument in a `Dense` layer within a `Sequential` model. To address this, I will explicitly add an `Input` layer as the first layer of the model.



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

# 1. Instantiate a Sequential model
model = Sequential()

# Add an Input layer explicitly to resolve the UserWarning
model.add(Input(shape=(num_pixels,)))

# 2. Add a Dense layer as the input layer (now implicitly connected to Input layer)
model.add(Dense(128, activation='relu'))

# 3. Add at least one more Dense hidden layer
model.add(Dense(64, activation='relu'))

# 4. Add a final Dense output layer
# num_classes was defined in the previous step as 10
model.add(Dense(num_classes, activation='softmax'))

# Print the model summary to confirm its structure
print("Neural Network Model Summary:")
model.summary()

print("Neural network model constructed successfully.")
```

## Compile Neural Network Model

### Subtask:
Compile the neural network model using `categorical_crossentropy` as the loss function, `Adam` optimizer, and `accuracy` as the metric.

**Reasoning**:
Following the subtask instructions, I will compile the neural network model using the specified loss function, optimizer, and metrics.



```python
from tensorflow.keras.optimizers import Adam

# Compile the model
model.compile(
    loss='categorical_crossentropy',
    optimizer=Adam(), # Using Adam optimizer as it's a common and effective choice
    metrics=['accuracy']
)

print("Neural network model compiled successfully with Adam optimizer, categorical_crossentropy loss, and accuracy metric.")
```

## Train Neural Network Model

### Subtask:
Train the neural network model using the preprocessed training data. Utilize appropriate hyperparameters like epochs and batch size, and ensure accuracy is tracked as a performance metric during training.

**Reasoning**:
Now that the model has been defined and compiled, I will train the neural network using the `model.fit()` method. I will use the preprocessed `x_train` and `y_train` for training, and `x_test` and `y_test` for validation to monitor performance and potential overfitting, along with setting `epochs` and `batch_size`.



```python
print("Training the neural network model...")

# Define training hyperparameters
epochs = 10
batch_size = 128

# Train the model
history = model.fit(
    x_train,
    y_train,
    epochs=epochs,
    batch_size=batch_size,
    validation_data=(x_test, y_test),
    verbose=1
)

print(f"Neural network model trained for {epochs} epochs with a batch size of {batch_size}.")
```

## Evaluate Neural Network Model

### Subtask:
Evaluate the performance of the trained model using the testing dataset, ensuring that the model generalizes well to unseen data. This includes reporting the loss and accuracy on the test set.

**Reasoning**:
To evaluate the trained model on the unseen test data, I will use the `model.evaluate()` method with the `x_test` and `y_test` datasets.



```python
print("Evaluating the neural network model on the test data...")

loss, accuracy = model.evaluate(x_test, y_test, verbose=0)

print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
print("Model evaluation complete.")
```

## Implement Prediction Functionality

### Subtask:
Develop a function to make predictions on new, unseen single images or batches of images, returning the predicted digit class with the highest probability.


**Reasoning**:
To implement the prediction functionality, I will first select a few sample images from the `x_test` dataset and their corresponding true labels. Then, I will ensure these samples are reshaped to match the model's expected input format (flattened 784 pixels). After that, I will use the trained model to make predictions and interpret the probability distributions to determine the predicted digit class for each sample. Finally, I will print both the predicted and true digits for each sample to evaluate the model's performance on individual examples.

**Reasoning**:
Following the reasoning, I will now implement the prediction functionality by selecting sample images, reshaping them if necessary, making predictions with the trained model, and then printing the true and predicted labels.



```python
import numpy as np

print("Implementing prediction functionality...")

# 1. Select a few sample images from x_test
# Let's pick the first 5 images for demonstration
sample_indices = range(5)
sample_images = x_test[sample_indices]
sample_true_labels = y_test[sample_indices]

# 2. Reshape these sample images to be compatible with the model's input
# (Already flattened during preprocessing, so this step is for clarity/reassurance)
# Model expects (num_samples, num_pixels)
# x_test is already (10000, 784), so no reshaping needed here for these specific samples.
# If we were to take raw 28x28 images, we'd do:
# sample_images_reshaped = sample_images.reshape(sample_images.shape[0], num_pixels)

# 3. Use the trained model to make predictions on these sample images
predictions = model.predict(sample_images)

# 4. Interpret the model's output to determine the predicted digit class
predicted_classes = np.argmax(predictions, axis=1)
true_classes = np.argmax(sample_true_labels, axis=1)

# 5. Print the predicted digit along with the true digit for each sample
print("\nPredictions for sample images:")
for i in range(len(sample_indices)):
    print(f"Sample {i+1}: True Digit: {true_classes[i]}, Predicted Digit: {predicted_classes[i]}, Probabilities: {predictions[i]}")

print("Prediction functionality implemented and tested on sample images.")
```
