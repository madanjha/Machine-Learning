##**Foundation**
###**What is Optimization**
The main goal of machine learning is to create a model that generalize well to the new or unseen data.

**Optimization is the process of finding the best possible model which help us to achive our ML Goal**

###**Parameter**
The things that model learns from the data

* **Let's say i am cooking Soup, I will be adding the salt according to my taste** -> Here the salt value is what **`Parameter`**
    * Technically -> The Coefficients in linear regression (m,c) and the weights in a neural network (y = wx + b). Thse are the parameters that are learned during the `.fit()` process

###**Hyperparameters**
Setting you tell the model before it starts learning is hyperparameter
* The Recipe instructions we set before cooking (temperature of oven, timer -> 30 mins)
    * Techinically -> The `n_estimators`(number of tree) in a Random Forest, The `C Parameter` (regularization strength) in support Vector Machine, In Gradient Descent we have `learning_rate`

**The Problem That we need to handle**
###**Overfitting**
When the model is too complex. It Learns the training data and it's noise aswell. it fails to generalize the new data (High Variance)
* Let's say we are memorizing the answers for the practice exam but not learning the concepts, so in Real exam you can fail

###**Underfitting**
**When the model is too simle and fails to learn the hidden patterns in trainin data. (High Bias)**


**Note:** Hyperparameter tuning is the process of finding the best settings to create a nodel that is `'Good'`

###**How we measure **GOOD** : Cross validation**
We cannot use the test set of the dataset to tune our hyperparamter. As it is like a cheating for getting good results on my testing data

**We use K-fold Cross Validation**
1. Split the training data into 'k' folds (k = 5)
2. Train the model on 4 folds, validates on the 1 held-out fold
3. Repeate 5 times, holding out a different folds each time
4. The score to set the values of hyperparmeter is the average score of all 5 folds

It gives us a good estiamte of how the model will perform on unseen data

##**Core Tuning Methods**
###1. GridSearchCV (Exhaustive Search)
We define a grid with all possible hyperparameter combinations that we want to try

**How it Works**
* It trains and evaluates a model for every single combinations in the grid using Cross validation
* pros: Very easy for understanding and it gives gaurantee to find the best combinations within your grid
* cons: It is extremly slow and suffers from the 'Curse of Dimensionality'.
if you have 3 parameters with 10 values each, `10 * 10 * 10 = 1000` Models to train (times of k folds)

###2. RandomizedSearchCV (Smart Sampling)
We will define a 'distribution' or 'list' for each hyperparameters. Then it will sample a fixed number(n_iter) of random combination from these distributions

**How it Works**
* let's say we gave `n_iter = 50`
* It will randomly picks 50 different combinations
* train them and report the best one

**Pros**
* Much Faster
* Not all hyperparameters are equally important. Random Search is most likely to find 'Good Enough' value for the important parameters, rather than wasitng time tesing many 'bad' values for unimportant parameters

**Cons**
* It does not guaranteed to find the best combinations

**To implement RandomSearch CV and Grid Search CV**

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from scipy.stats import uniform, loguniform # For random search distribution
```

**We will be using the same Scaled Training Data from the above code**

```python
x_train_scaled , y_train
```

**GridSearchCV**

```python
svc = SVC(random_state=42)
```

**Define the 'grid' of hyperparameter to test**

```python
param_grid = {
    'C': [0.1,1,10,100],
    'gamma': [1,0.1,0.01,0.001],
    'kernel': ['rbf','sigmoid']
}
# 4 (C) * 4 (gamma) * 2 (Kernals) * 5 (CV) = 160 models
```

**Now, we will setup the gridsearchCV**

```python
grid_search = GridSearchCV(estimator=svc,
                           param_grid = param_grid,
                           cv = 5, # We are creating five folds in training data for validation
                           n_jobs = -1, # It means we are using all the available CPU cores
                           verbose = 1) # verbose shows the progess
```

**Run the search**

```python
grid_search.fit(x_train_scaled,y_train)
```

```python
print(f'GridSearch Best Score (CV Accuracy): {grid_search.best_score_:.2f}')
print(f'GridSeach Best Params: {grid_search.best_params_}')
```

**RandomSearchCV**

First, we will define the 'distributions' to sample from. It will give us much wider range to explore


```python
param_dist = {
    'C': loguniform(1e-1, 1e2), # sample from 0.1 to 100
    'gamma': loguniform(1e-3,1e0), # sample from 0.001 to 1
    'kernel': ['rbf','sigmoid']
}
```

**Now we will setup our RandomSearchCV**
* n_iter = 50 --> It will try 50 random combination
* total models trained = 50 (n_iter) * 5(cv) = 250 models
* This is more than GridSearchCV, but explores a **much** wider space

```python
random_search = RandomizedSearchCV(estimator=svc,
                                   param_distributions = param_dist,
                                   n_iter = 50,
                                   cv = 5,
                                   n_jobs = -1,
                                   verbose = 1,
                                   random_state = 42)
```

Run the search

```python
random_search.fit(x_train_scaled,y_train)
```

```python
print(f'RandomSearch Best Score (CV Accuracy): {random_search.best_score_:.2f}')
print(f'RandomSearch Best Params: {random_search.best_params_}')
```

**Final Evalutation**

```python
best_model = grid_search.best_estimator_ # or random_search.best_estimator_
```

```python
final_y_pred = best_model.predict(x_test_scaled)
```

```python
final_y_pred
```

```python
final_accuracy = accuracy_score(y_test,final_y_pred)
```

```python
print(f'BaseModel Test Accuracy: {test_accuracy:.4f}')
print(f'Tuned Test Accuracy: {final_accuracy:.4f}')
```

##**Other Optimization Techniques**
Hyperparameter tuning is not the only wat to optimize the model. We also optimize the model's parameter and structure

###**Gradient Descent**
This is the algorithm that perfoms the leaning(finds the paramater) for most models like linear Regression, logistic Regression, Neural Network

**Goal** - To minimize the Loss function **(`J(0)` -> It is basically a mathematic function that measures how wrong the model predictions are)**

###**Regularization**
It is a technique to prevent Overfitting by penalizing the complex model.

`cost = loss(how wrong we are) + lambda * Penality(How complex we are)`
* lambda -> The regularization strength

**It has two types based on penalizing way**
1. L2 Regularization (Rigid):
    * Penality is the sum of the squared weights
    * It basically shrinks the weights but not make it 0

2. L1 Regularization (Lasso):
    * Penality is the sum of absolute values of the weights
    * It forces many weights to be exaclty zero
    * It automatically does feature selection, if weight of feature is zero -> unimportant

```python
from sklearn.linear_model import LogisticRegression, SGDClassifier
```

**Logistic Reg has builtin regularization**
* 'C' is the 'inverse' of regularization strenght (small C = Strong Regularization)

```python
log_reg = LogisticRegression(solver='liblinear',random_state=42, max_iter=1000)
```

```python
reg_para_grid = {
    'C' : [0.01,0.1,1,10,100],
    'penalty' : ['l1','l2']
}
```

```python
reg_grid_seach = GridSearchCV(log_reg,
                              reg_para_grid,
                              cv = 5,
                              n_jobs = -1,
                              verbose = 1)
```

```python
reg_grid_seach.fit(x_train_scaled,y_train)
```

```python
print(f'Best Regularization Score: {reg_grid_seach.best_score_:.2f}')
print(f'Best Regularization params: {reg_grid_seach.best_params_}')
```

**Now I will build SGD model**

```python
sgd_model = SGDClassifier(loss = 'log_loss',
                          penalty='l2',
                          max_iter = 1000,
                          tol = 1e-3)
```

```python
sgd_model.fit(x_train_scaled,y_train)
```

```python
y_pred_sgd = sgd_model.predict(x_test_scaled)
```

```python
accuracy_score(y_test,y_pred_sgd)
```
