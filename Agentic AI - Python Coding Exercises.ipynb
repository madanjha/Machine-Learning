```python
class ReActAgent:
    def __init__(self):
        # Indented 4 spaces: belongs to the class
        self.thought_log = []

    def think(self, observation):
        """ Reason about what to do """
        # Indented 4 spaces: belongs to the class
        # Indented 8 spaces: belongs to the method
        thought = f"Observation: {observation}."
        
        if "?" in observation:
            thought += " Thought: This is a question, I should search."
            action = "search" # Logic to return action
        else:
            thought += " Thought: This is a statement, I should acknowledge."
            action = "acknowledge" # Logic to return action
            
        return action, thought

    def react(self, observation):
        """ Complete ReAct cycle """
        # Receive both the action and the reasoning from the think method
        action, thought = self.think(observation)
        
        # Store the history of thoughts and actions
        self.thought_log.append((observation, thought, action))
        
        return f"Action Taken: {action} | Reasoning: {thought}"

# --- Test (No indentation: this is the main execution block) ---
agent = ReActAgent()
print(agent.react("What is AI?"))
print(agent.react("The sky is blue."))
```

```python
class MemoryAgent:
    def __init__(self, max_memory=5):
        self.max_memory = max_memory
        self.conversation_history = []

    def remember(self, user_input, agent_response):
        """ Store conversation in memory """
        # Create a dictionary for the current turn
        entry = {'user': user_input, 'agent': agent_response}
        self.conversation_history.append(entry)

        # Keep only last max_memory items (Sliding Window Memory)
        if len(self.conversation_history) > self.max_memory:
            self.conversation_history.pop(0)

    def recall(self):
        """ Get conversation context """
        context = ""
        for conv in self.conversation_history:
            context += f"User: {conv['user']}\nAgent: {conv['agent']}\n"
        return context

# --- Test (Unindented to run as main script) ---
agent = MemoryAgent(max_memory=3)
agent.remember("Hi", "Hello!")
agent.remember("How are you?", "I’m good!")

print(agent.recall())
```

```python
class TaskDecompositionAgent:
    def decompose(self, task):
        """ Break task into subtasks """
        subtasks = []
        
        # Split task by "and" or " then " keywords
        if " and " in task.lower():
            subtasks = task.lower().split(" and ")
        elif " then " in task.lower():
            subtasks = task.lower().split(" then ")
        else:
            subtasks = [task]

        return subtasks

    def execute_plan(self, task):
        """ Execute decomposed tasks """
        subtasks = self.decompose(task)
        results = []

        # Execute each subtask
        for subtask in subtasks:
            result = f"Completed: {subtask.strip()}"
            results.append(result)

        return results

# --- Test ---
agent = TaskDecompositionAgent()
plan = "Buy groceries and cook dinner"
print(agent.execute_plan(plan))
```

```python
class SelfReflectiveAgent:
    def generate_response(self, query):
        """ Initial response """
        return f"Response to: {query}"

    def critique(self, response):
        """ Evaluate own response """
        if len(response) < 20:  # Increased threshold to trigger for "Response to: Hi"
            return "Response is too short"
        elif "ERROR" in response:
            return "Response contains an ERROR"
        else:
            return "Response looks good"

    def improve(self, query, response, critique):
        """ Refine response based on critique """
        if "too short" in critique.lower():
            return f"{response}. I should provide more detail to be helpful."
        else:
            return response

# --- Test ---
agent = SelfReflectiveAgent()
resp = agent.generate_response("Hi")
critique = agent.critique(resp)
improved = agent.improve("Hi", resp, critique)

print(f"Original: {resp}")
print(f"Critique: {critique}")
print(f"Improved: {improved}")
```

```python
class CollaborativeAgent:
    def __init__(self, name, specialty):
        # Indented inside __init__
        self.name = name
        self.specialty = specialty
        self.inbox = []

    def send_message(self, recipient, message):
        """ Send message to another agent """
        # Logic to add message to recipient's inbox
        recipient.inbox.append(message)

    def receive_messages(self):
        """ Process all messages """
        responses = []
        # Logic to process each message in inbox
        for msg in self.inbox:
            response = f"{self.name} received: {msg}"
            responses.append(response)

        # Clear inbox after processing
        self.inbox.clear()
        return responses

# --- Test ---
agent1 = CollaborativeAgent("Agent1", "research")
agent2 = CollaborativeAgent("Agent2", "coding")

# Agent1 sends a message to Agent2
agent1.send_message(agent2, "Need help with code")

# Agent2 checks their inbox
print(agent2.receive_messages())
```
